```python
import numpy as np
import pandas as pd
```

```python
class Linear_regressor:
    def __init__(self,learning_rate=0.01, iteration=1000):
        self.learning_rate = learning_rate
        self.iteration = iteration
        self.m = None
        self.b = None 
        self.losses = []

    def fit(self,X,Y):
        n=len(X)
        self.m = 0
        self.b = 0

        for i in range(self.iteration):
            y_pred = self.m*X +self.b

            loss = (1/n)* np.sum((Y-y_pred)**2)
            self.losses.append(loss)

            dm = (-2/n)* np.sum(X*(Y-y_pred))
            db = (-2/n)* np.sum(Y-y_pred)

            self.m = self.m - self.learning_rate*dm
            self.b = self.b - self.learning_rate*db

            if i%100 == 0:
                print(f"iteration {i}, loss: {loss:.4f}")
    def predict(self, X):
        return self.m*X + self.b
    
    def score(self,X,Y):
        y_pred = self.predict(X)
        ss_res = np.sum((Y-y_pred)**2)
        ss_tot = np.sum((Y-np.mean(Y))**2)
        return 1 - (ss_res/ss_tot)

```

```python
df = pd.read_csv("placement.csv")
X = df["cgpa"].values
Y = df["placement_exam_marks"].values
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,Y,test_size=0.2,random_state=2)
```

```python
modal = Linear_regressor(learning_rate=0.01, iteration=1000)
```

```python
modal.fit(X_train,y_train)
```

```python
modal.predict(X_test)
```

```python
modal.score(X_test,y_test)
```
